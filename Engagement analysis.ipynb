```python
def check_duplicate(dataset):
    """
    Counts the number of duplicate rows in the dataset.

    Parameters:
    dataset (pd.DataFrame): The dataset to check for duplicate rows.

    Returns:
    int: The number of duplicate rows in the dataset.


    """

    duplicate_count=dataset.duplicated().sum()
    return duplicate_count  

duplicate_count=check_duplicate(df)
print(f'There are {duplicate_count} duplicate rows in the dataset.')
```

```python
def get_categorical_columns(dataset):
    """
    Retrieves categorical columns from the dataset.

    Parameters:
    dataset (pd.DataFrame): The dataset from which to retrieve categorical columns.

    Returns:
    pd.DataFrame: A DataFrame containing only the categorical columns.

    
    """
    categorical_columns = dataset.select_dtypes(include=['object', 'category'])
    unique_values_cat_cols=categorical_columns.apply(lambda x: x.unique())
    return unique_values_cat_cols
unique_values_count = get_categorical_columns(df)
print(unique_values_count)

```

# Temporal Analysis

```python
def time_date_split(dataset, col):
    """
    Splits the given datetime column into separate date and time columns, 
    and adds month and weekday names.

    Parameters:
    dataset (pd.DataFrame): The DataFrame containing the data.
    col (str): The column name of the datetime field to be split.

    The function adds the following new columns to the dataset:
    - 'Date': The date part extracted from the datetime column.
    - 'Time': The time part, rounded to the nearest hour.
    - 'Month_name': The name of the month.
    - 'Weekday_name': The name of the weekday.
    
    
    """
    dataset['Date'] = pd.to_datetime(dataset[col]).dt.date
    dataset['Time'] = pd.to_datetime(dataset[col]).dt.round('h').dt.time
    dataset['Weekend and Weekday'] = pd.to_datetime(dataset[col]).dt.day_name()
    df['Hour'] = pd.to_datetime(dataset[col]).dt.hour
    df['Year']=pd.to_datetime(dataset[col]).dt.year


time_date_split(df,'Post Timestamp')

 
```

```python
def categorize_time(time):
    
    """
    Categorizes a given time into one of four time periods: Morning, Afternoon, Evening, or Night.
    
    Parameters:
    time (datetime.time): The time to be categorized.
    
    Returns:
    str: The category of the time period ('Morning', 'Afternoon', 'Evening', or 'Night').
    
    Time periods are defined as:
    - Morning: 06:00:00 to 11:59:59
    - Afternoon: 12:00:00 to 16:59:59
    - Evening: 17:00:00 to 20:59:59
    - Night: 21:00:00 to 05:59:59
    """
    if datetime.strptime('06:00:00', '%H:%M:%S').time() <= time < datetime.strptime('12:00:00', '%H:%M:%S').time():
        return 'Morning'
    elif datetime.strptime('12:00:00', '%H:%M:%S').time() <= time < datetime.strptime('17:00:00', '%H:%M:%S').time():
        return 'Afternoon'
    elif datetime.strptime('17:00:00', '%H:%M:%S').time() <= time < datetime.strptime('21:00:00', '%H:%M:%S').time():
        return 'Evening'
    else:
        return 'Night'

df['time_of_day'] = df['Time'].apply(categorize_time) 
```

# Geographical Analysis

```python
url='https://statisticstimes.com/geography/countries-by-continents.php'
```

```python
tables=pd.read_html(url)
continent_data=tables[2]
continent_data.head(10)
```

```python
continent_data_cleaned=continent_data[['Country or Area','Continent']].copy()
continent_data_cleaned.columns=['Audience Location', 'Continent']
continent_data_cleaned.dropna(inplace=True)
merged_data = pd.merge(df, continent_data_cleaned, on='Audience Location', how='left')
merged_data.head(10)
```

>The five continents with the highest engagement rates

```python
result = merged_data.groupby('Continent')['Engagement Rate'].mean().reset_index().nlargest(5, 'Engagement Rate')

plt.figure(figsize=(12, 8))  # Correct figure size
result.plot(kind='bar', x='Continent', y='Engagement Rate', legend=False, color='skyblue')

# Adding labels and title
plt.title('Top 5 Continents with the Highest Engagement Rates')
plt.xlabel('Continent')
plt.ylabel('Engagement Rate')

# Show the plot
plt.show()
```
